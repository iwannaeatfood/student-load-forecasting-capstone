```python
from upp_forecast.preprocessing.build_cohorts import (
    COHORT_COLUMNS,
    MEASURE_COLUMNS,
    STATE_COLUMNS,
    build_cohorts,
)
from upp_forecast.preprocessing.clean_terms import TERM_SCHEMA, clean_terms


time_cols = [
    TERM_SCHEMA.academic_year_col,
    TERM_SCHEMA.term_col,
]

required_actuals_cols = (
    COHORT_COLUMNS
    + time_cols
    + STATE_COLUMNS
    + MEASURE_COLUMNS
)
missing_cols = sorted(set(required_actuals_cols) - set(actuals_data.columns))
if missing_cols:
    raise KeyError(f"Actuals Data.csv is missing required columns: {missing_cols}")

required_actuals_cols
```

```python
actuals_for_state = actuals_data[required_actuals_cols].copy()
records = clean_terms(actuals_for_state)
```

```python
cohort_state_table = build_cohorts(records)

cohort_state_table.head(20)
```

```python
cohort_state_table.to_csv(project_root / "data" / "processed" / "cohort_state_table.csv", index=False)
```
